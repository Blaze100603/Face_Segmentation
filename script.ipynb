```python


import os
import json
import time
import cv2
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from sklearn.model_selection import train_test_split
from tensorflow.keras import layers, Model
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ReduceLROnPlateau,
    ModelCheckpoint
)
DATA_PATH = "train.npy"
IMG_SIZE = 256
BATCH_SIZE = 8
EPOCHS = 40
SEED = 42

np.random.seed(SEED)
tf.random.set_seed(SEED)

os.makedirs("models", exist_ok=True)

print("=" * 70)
print("FACE SEGMENTATION TRAINING PIPELINE")
print("=" * 70)

print("\nLoading dataset...")
raw_data = np.load(DATA_PATH, allow_pickle=True)

print("Raw dataset shape:", raw_data.shape)


def create_binary_mask(height, width, annotations):
    mask = np.zeros((height, width), dtype=np.uint8)

    if isinstance(annotations, list):
        for ann in annotations:
            points = ann.get("points", [])

            if len(points) >= 2:
                x1 = int(points[0]["x"] * width)
                y1 = int(points[0]["y"] * height)

                x2 = int(points[1]["x"] * width)
                y2 = int(points[1]["y"] * height)

                x1, x2 = sorted([x1, x2])
                y1, y2 = sorted([y1, y2])

                x1 = max(0, min(width - 1, x1))
                x2 = max(0, min(width - 1, x2))
                y1 = max(0, min(height - 1, y1))
                y2 = max(0, min(height - 1, y2))

                cv2.rectangle(mask, (x1, y1), (x2, y2), 1, -1)

    return mask

images = []
masks = []

print("\nProcessing dataset...")

for idx, sample in enumerate(raw_data):
    image = sample[0]
    annotations = sample[1]

    # Ensure RGB
    if image.ndim == 2:
        image = cv2.cvtColor(image, cv2.COLOR_GRAY2RGB)

    elif image.ndim == 3 and image.shape[2] == 4:
        image = cv2.cvtColor(image, cv2.COLOR_RGBA2RGB)

    image = cv2.resize(image, (IMG_SIZE, IMG_SIZE))
    mask = create_binary_mask(IMG_SIZE, IMG_SIZE, annotations)

    images.append(image)
    masks.append(mask)

    if (idx + 1) % 100 == 0:
        print(f"Processed {idx+1}/{len(raw_data)}")

images = np.array(images, dtype=np.uint8)
masks = np.array(masks, dtype=np.uint8)

print("\nImages shape:", images.shape)
print("Masks shape :", masks.shape)

print("\nShowing sample data...")

plt.figure(figsize=(12,4))
for i in range(3):
    idx = np.random.randint(0, len(images))

    plt.subplot(2,3,i+1)
    plt.imshow(images[idx])
    plt.title("Image")

    plt.subplot(2,3,i+4)
    plt.imshow(masks[idx], cmap="gray")
    plt.title("Mask")

plt.tight_layout()
plt.show()

X = images.astype(np.float32) / 255.0
Y = masks.astype(np.float32)
Y = np.expand_dims(Y, axis=-1)

X_train, X_val, y_train, y_val = train_test_split(
    X, Y,
    test_size=0.2,
    random_state=SEED
)

print("\nTrain:", X_train.shape)
print("Val  :", X_val.shape)

def dice_coefficient(y_true, y_pred, smooth=1e-6):
    y_true_f = tf.reshape(y_true, [-1])
    y_pred_f = tf.reshape(y_pred, [-1])

    intersection = tf.reduce_sum(y_true_f * y_pred_f)

    return (2. * intersection + smooth) / (
        tf.reduce_sum(y_true_f) +
        tf.reduce_sum(y_pred_f) +
        smooth
    )

def iou_metric(y_true, y_pred, smooth=1e-6):
    y_pred = tf.cast(y_pred > 0.5, tf.float32)

    intersection = tf.reduce_sum(y_true * y_pred)
    union = tf.reduce_sum(y_true) + tf.reduce_sum(y_pred) - intersection

    return (intersection + smooth) / (union + smooth)

def f1_score_metric(y_true, y_pred):
    y_pred = tf.cast(y_pred > 0.5, tf.float32)

    tp = tf.reduce_sum(y_true * y_pred)
    fp = tf.reduce_sum((1-y_true) * y_pred)
    fn = tf.reduce_sum(y_true * (1-y_pred))

    precision = tp / (tp + fp + 1e-6)
    recall = tp / (tp + fn + 1e-6)

    return 2 * precision * recall / (precision + recall + 1e-6)

def combo_loss(y_true, y_pred):
    bce = tf.keras.losses.binary_crossentropy(y_true, y_pred)
    dice = 1 - dice_coefficient(y_true, y_pred)

    return 0.5 * tf.reduce_mean(bce) + 0.5 * dice


def conv_block(x, filters):
    x = layers.Conv2D(filters, 3, padding="same", activation="relu")(x)
    x = layers.Conv2D(filters, 3, padding="same", activation="relu")(x)
    return x

def encoder_block(x, filters):
    c = conv_block(x, filters)
    p = layers.MaxPooling2D()(c)
    return c, p

def decoder_block(x, skip, filters):
    x = layers.UpSampling2D()(x)
    x = layers.Concatenate()([x, skip])
    x = conv_block(x, filters)
    return x

def build_unet():
    inputs = layers.Input((IMG_SIZE, IMG_SIZE, 3))

    s1, p1 = encoder_block(inputs, 32)
    s2, p2 = encoder_block(p1, 64)
    s3, p3 = encoder_block(p2, 128)
    s4, p4 = encoder_block(p3, 256)

    b1 = conv_block(p4, 512)

    d1 = decoder_block(b1, s4, 256)
    d2 = decoder_block(d1, s3, 128)
    d3 = decoder_block(d2, s2, 64)
    d4 = decoder_block(d3, s1, 32)

    outputs = layers.Conv2D(1, 1, activation="sigmoid")(d4)

    model = Model(inputs, outputs, name="U-Net")

    return model

model = build_unet()

model.compile(
    optimizer=tf.keras.optimizers.Adam(1e-4),
    loss=combo_loss,
    metrics=[
        dice_coefficient,
        iou_metric,
        f1_score_metric
    ]
)

model.summary()

callbacks = [
    EarlyStopping(
        monitor="val_dice_coefficient",
        mode="max",
        patience=8,
        restore_best_weights=True,
        verbose=1
    ),

    ReduceLROnPlateau(
        monitor="val_dice_coefficient",
        mode="max",
        factor=0.5,
        patience=4,
        verbose=1
    ),

    ModelCheckpoint(
        "models/unet_mobilenetv2_best_weights.h5",
        monitor="val_dice_coefficient",
        mode="max",
        save_best_only=True,
        verbose=1
    )
]

print("\nStarting training...\n")

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    batch_size=BATCH_SIZE,
    epochs=EPOCHS,
    callbacks=callbacks,
    verbose=1
)

print("\nEvaluating model...")

results = model.evaluate(X_val, y_val, verbose=0)

loss_val = float(results[0])
dice_val = float(results[1])
iou_val = float(results[2])
f1_val = float(results[3])

print(f"Loss : {loss_val:.4f}")
print(f"Dice : {dice_val:.4f}")
print(f"IoU  : {iou_val:.4f}")
print(f"F1   : {f1_val:.4f}")
sample = X_val[:1]

start = time.time()
_ = model.predict(sample, verbose=0)
infer_ms = (time.time() - start) * 1000

model.save("models/unet_mobilenetv2_model.h5")
model.save("models/unet_mobilenetv2_model.keras")

config = {
    "model_name": "U-Net Face Segmentation",
    "input_shape": [256,256,3],
    "output_shape": [256,256,1],

    "performance_metrics": {
        "dice_coefficient": dice_val,
        "iou": iou_val,
        "f1_score": f1_val,
        "inference_time_ms": infer_ms
    }
}

with open("models/deployment_config.json", "w") as f:
    json.dump(config, f, indent=4)

print("\nSaved:")
print("models/unet_mobilenetv2_model.h5")
print("models/unet_mobilenetv2_model.keras")
print("models/deployment_config.json")

print("\nPrediction Samples...")

preds = model.predict(X_val[:3], verbose=0)

plt.figure(figsize=(12,9))

for i in range(3):
    plt.subplot(3,3,i*3+1)
    plt.imshow(X_val[i])
    plt.title("Image")

    plt.subplot(3,3,i*3+2)
    plt.imshow(y_val[i].squeeze(), cmap="gray")
    plt.title("True Mask")

    plt.subplot(3,3,i*3+3)
    plt.imshow((preds[i].squeeze()>0.5), cmap="gray")
    plt.title("Predicted")

plt.tight_layout()
plt.show()

print("\nTraining complete.")
```
